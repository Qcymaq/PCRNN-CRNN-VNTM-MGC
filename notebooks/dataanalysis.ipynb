# For data 10  seconds

```python
input_data_folder = os.path.join(DATA_DIR, "VNTM3")  
output_data_folder = os.path.join(DATA_DIR, "dataset10seconds") 

for folder_name in os.listdir(input_data_folder):
    output_folder = os.path.join(output_data_folder, folder_name)
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

for folder_name in os.listdir(input_data_folder):
    input_folder = os.path.join(input_data_folder, folder_name)
    output_folder = os.path.join(output_data_folder, folder_name)
    utils.split_audio_in_folder(input_folder, output_folder)
```
